# 📐 Geometric Interpretation of Linear Systems & Matrix Ranks

## 📌 Overview
This notebook visually demonstrates the three fundamental scenarios of systems of linear equations ($Ax = b$) in a 3D space:
1. **Unique Solution** ($\text{Rank}(A) = \text{Rank}(A|b) = 3$)
2. **No Solution** ($\text{Rank}(A) \neq \text{Rank}(A|b)$ due to parallel planes)
3. **Infinitely Many Solutions** ($\text{Rank}(A) = \text{Rank}(A|b) < 3$)

Developed as part of the **Linear Algebra & Optimization** coursework for M.Tech Data Science at M.S. Ramaiah Institute of Technology.


```python
import numpy as np
import matplotlib.pyplot as plt

# Create a grid of points for X and Y coordinates
x = np.linspace(-5, 5, 20)
y = np.linspace(-5, 5, 20)
X, Y = np.meshgrid(x, y)

# Initialize figure with 3 subplots side-by-side
fig = plt.figure(figsize=(18, 6))

# --- Case 1: Unique Solution (Planes intersect at a single point) ---
ax1 = fig.add_subplot(131, projection='3d')
Z1 = X + Y
Z2 = 2*X - Y + 1
Z3 = -X + 3*Y - 2
ax1.plot_surface(X, Y, Z1, alpha=0.5, color='red', label='Plane 1')
ax1.plot_surface(X, Y, Z2, alpha=0.5, color='green', label='Plane 2')
ax1.plot_surface(X, Y, Z3, alpha=0.5, color='blue', label='Plane 3')
ax1.set_title("1. Unique Solution\nRank(A) = Rank(A|b) = 3", fontsize=11, fontweight='bold')
ax1.set_xlabel('X axis')
ax1.set_ylabel('Y axis')
ax1.set_zlabel('Z axis')

# --- Case 2: No Solution (Strictly parallel planes) ---
ax2 = fig.add_subplot(132, projection='3d')
Z4 = X + Y
Z5 = X + Y + 4
Z6 = X + Y - 4
ax2.plot_surface(X, Y, Z4, alpha=0.5, color='red')
ax2.plot_surface(X, Y, Z5, alpha=0.5, color='green')
ax2.plot_surface(X, Y, Z6, alpha=0.5, color='blue')
ax2.set_title("2. No Solution\nRank(A) != Rank(A|b)", fontsize=11, fontweight='bold')
ax2.set_xlabel('X axis')
ax2.set_ylabel('Y axis')
ax2.set_zlabel('Z axis')

# --- Case 3: Infinitely Many Solutions (Coincident or intersecting along a line) ---
ax3 = fig.add_subplot(133, projection='3d')
Z7 = X + Y
Z8 = 2*X + 2*Y
Z9 = 0.5*X + 0.5*Y
ax3.plot_surface(X, Y, Z7, alpha=0.5, color='red')
ax3.plot_surface(X, Y, Z8, alpha=0.5, color='green')
ax3.plot_surface(X, Y, Z9, alpha=0.5, color='blue')
ax3.set_title("3. Infinite Solutions\nRank(A) = Rank(A|b) < 3", fontsize=11, fontweight='bold')
ax3.set_xlabel('X axis')
ax3.set_ylabel('Y axis')
ax3.set_zlabel('Z axis')

plt.tight_layout()

# Save high-resolution graphic for README embedding
plt.savefig("3D_Linear_Systems_Assignment.png", dpi=300, bbox_inches='tight')
plt.show()
```

## 🧠 Key Theoretical Takeaways
- **Row Echelon Form (REF):** Systems can be evaluated using Gaussian elimination to determine matrix rank.
- **Geometric Insight:**
  - When three non-parallel planes intersect cleanly, they share a single coordinate triplet $(x, y, z)$[cite: 1].
  - When planes share the exact same slope vector but different intercepts, they never meet, leaving the system inconsistent[cite: 1].
  - Linearly dependent equations collapse the system dimension, producing a continuous line or plane of valid solutions[cite: 1].
